### Filter for harmonies

```python
harmonies = pitchclass[pitchclass['pc'].apply(lambda x: len(x) > 1)]
harmonies
```

```python
dissonance_table = {
    0: 0.0,    # Unison
    1: 1.0,    # m2/M7
    2: 0.6,    # M2/m7
    3: 0.4,    # m3/M6
    4: 0.2,    # M3/m6
    5: 0.0,    # P4/P5
    6: 0.8,    # Tritone
}
```

```python
def preprocess_df(df_raw):
    df = df_raw.copy()
    df['pc'] = df['midi'] % 12
    df['note'] = df['pc'].map(lambda x: pitch_class_names[x])
    
    df['mc_onset_float'] = df['mc_onset'].apply(lambda x: float(Fraction(x)))
    df['order'] = df['mc'] + df['mc_onset_float']

    return df
```

```python
def generate_pc(df_raw):
    pitchclass = df_raw.copy()
    pitchclass = pitchclass.groupby('order')[['note', 'pc']].agg(list).reset_index()
    return pitchclass
```

```python
import itertools

def interval_class(lst):
    output = []
    for a, b in itertools.combinations(lst, 2):
        interval = abs(a - b) % 12  # if you're working mod 12
        IC = min(interval, 12 - interval)
        output.append(IC)
    return output
```

```python
path = "/Users/Jessica/Documents/ERIP2025/Jessica-ERIP2025/jazz_transcriptions/notes/Vulfpeck-Fugue_State.tsv"
df = pd.read_csv(path, sep='\t')
df = preprocess_df(df)
pc = generate_pc(df)
```

```python
df
```

```python
pc
```

```python
harmonies = pc[pc['pc'].apply(lambda x: len(x) > 1)].copy()
harmonies['interval'] = harmonies['pc'].apply(interval_class)
harmonies['num notes'] = harmonies['pc'].apply(lambda x: len(x))
harmonies['dissonance'] = (harmonies['interval'].apply(lambda x: sum(dissonance_table[i] for i in x))) / harmonies['num notes']
harmonies
```

```python
harmonies[harmonies['dissonance']> 1]
```

```python
x = harmonies['order']
y = harmonies['dissonance']
mean = harmonies['dissonance'].mean()
N = harmonies['order'].shape[0]
plt.figure(figsize=(10, 6))
plt.scatter(x, y)
plt.plot(x, np.zeros(N)+mean, label = 'Mean Dissonance Index', c='orange')
plt.xlabel('Measure')
plt.ylabel('Dissonance Index')
plt.title('Dissonance Index of Harmonies in Fugue State by Vulfpeck')
plt.legend()
plt.show
```
