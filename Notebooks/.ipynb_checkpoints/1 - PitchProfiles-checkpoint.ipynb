### Plot pitch profiles by standard

```python
title_grouped, ref = g.groupby_standard(df, length=3)
ref
```

```python
subset = df[(df['workTitle'] == 'Four') | (df['workTitle'] == 'Blues By Five')]
subset = subset[subset['artist'].isin(['John Coltrane', 'Miles Davis'])]

subset = subset.groupby(['artist', 'workTitle'], as_index=False).agg(list)
subset
```

```python
data = {}
songs = []
artists = []

for row in range(subset.shape[0]):
    subset_row = subset.iloc[[row]]
    worktitle = subset_row.iloc[0]['workTitle']
    
    if worktitle not in songs:
        songs.append(worktitle)
    
    
    artist = (subset_row.iloc[0]['artist'])

    if artist not in artists:
        artists.append(artist)

    key = (worktitle, artist)
    data[key] = subset_row.iloc[0][['note', 'pc']]

print(songs, artists) 
```

```python
s = data[('Four', 'John Coltrane')]
pd.DataFrame({'note': s['note'], 'pc': s['pc']})
```

```python
# --- Plot ---
fig = plt.figure(figsize=(12, 9),constrained_layout=True)
subfigs = fig.subfigures(nrows=2, ncols=1)

for i, song in enumerate(songs):
    subfig = subfigs[i]
    subfig.suptitle(f'{song}', fontsize=14)
    
    axs = subfig.subplots(nrows=1, ncols=2)
    
    row_max = 0
    for j, artist in enumerate(artists):
        ax = axs[j]
        
        # Get the pitch profile data:
        pitch_data = data[(song, artist)]
        pitch_data = pd.DataFrame({'pc': pitch_data['pc']})
        pitch_counts = pitch_data.groupby(['pc']).size().reset_index(name='count')
        count = pitch_counts.sort_values('pc')

        
        # Plot histogram:
        ax.bar(count['pc'], count['count'], color='skyblue', edgecolor='black')
        
        # Set labels:
        ax.set_title(f'Artist: {artist}')
        ax.set_xlabel('Pitch')
        ax.set_ylabel('Count')

# Global figure title:
fig.suptitle('Pitch Profiles of 2 Songs by 2 Artists', fontsize=16)

plt.show()
```

```python
# --- Plot ---
fig = plt.figure(figsize=(12, 9),constrained_layout=True)
subfigs = fig.subfigures(nrows=2, ncols=1)

for i, song in enumerate(songs):
    subfig = subfigs[i]
    subfig.suptitle(f'{song}', fontsize=14)
    
    axs = subfig.subplots(nrows=1, ncols=2)
    
    row_max = 0
    for j, artist in enumerate(artists):
        ax = axs[j]
        
        # Get the pitch profile data:
        pitch_data = data[(song, artist)]
        pitch_data = pd.DataFrame({'note': pitch_data['note'], 'pc': pitch_data['pc']})
        pitch_counts = pitch_data.groupby(['note', 'pc']).size().reset_index(name='count')
        count = pitch_counts.sort_values('pc')

        
        # Plot histogram:
        ax.bar(count['note'], count['count'], color='skyblue', edgecolor='black')
        
        # Set labels:
        ax.set_title(f'Artist: {artist}')
        ax.set_xlabel('Spelled Pitch')
        ax.set_ylabel('Count')

# Global figure title:
fig.suptitle('Pitch Profiles of 2 Songs by 2 Artists', fontsize=16)

plt.show()
```
