```python
metadata = g.load_metadata()
```

```python
bins = np.arange(1930, 2030, 10) 
years = [f"{start}-{start+10}" for start in bins[:-1]]
```

```python
num_scores = []
num_artists = []
for year in years:
    subset = metadata[metadata['year_bin'] == year]
    num_scores.append(subset.shape[0])
    num_artists.append(len(subset['composer'].unique()))
    
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].bar(years, num_scores, color='skyblue')
axes[0].set_title('Number of Scores per Decade')
axes[0].set_ylabel('Count')
axes[0].tick_params(axis='x', rotation=45)

axes[1].bar(years, num_artists, color='orange')
axes[1].set_title('Number of artists per Decade')
axes[1].set_ylabel('Count')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```
