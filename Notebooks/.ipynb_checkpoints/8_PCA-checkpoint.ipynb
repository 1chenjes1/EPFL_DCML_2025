```python
%reload_ext autoreload
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
from fractions import Fraction
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm
import music21 as m21
import umap
import hdbscan

import sys
sys.path.append('/Users/Jessica/Documents/ERIP2025/Jessica-ERIP2025/')

from Importables import General as g
from Importables import pitchanalysis_utils as pa
from Importables import clustering_utils as c
```

```python
metadata = g.load_metadata()
notes = g.load_notes(metadata)
notes_C = pa.transpose_to_C(notes)
```

```python
notes_vectors = c.get_clustering_data(notes_C)

subset_cols = [col for col in notes_vectors.columns if col != 'mc']
notes_no_dups = notes_vectors.drop_duplicates(subset=subset_cols)

bins = np.arange(1950, 2030, 10) 
bin_labels = [f"{start}-{start+10}" for start in bins[:-1]]

notes_no_dups = notes_no_dups[notes_no_dups['recording_year'].notna()].copy()
notes_no_dups['year_bin'] = pd.cut(notes_no_dups['recording_year'], bins=bins, labels=bin_labels, right=False)

notes_no_dups = notes_no_dups[notes_no_dups['year_bin'].notna()]
notes_no_dups
```

```python
from sklearn.preprocessing import StandardScaler

X = notes_no_dups.drop(columns=['mc', 'recording_year', 'year_bin', 'artist']).values

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

## 2 Component PCA

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Explained variance ratio:", pca.explained_variance_ratio_)
print("Total variance explained by 2 components:", sum(pca.explained_variance_ratio_))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

colors = plt.cm.get_cmap('tab10', len(bins))  

for i, year in enumerate(bin_labels):
    subset = X_pca[notes_no_dups['year_bin'] == year]
    
    n_sample = int(0.4 * len(subset))
    if n_sample > 0: 
        sampled_indices = np.random.choice(len(subset), n_sample, replace=False)
        subset = subset[sampled_indices]
    
    plt.scatter(subset[:,0], subset[:,1], color=colors(i), label=str(year), alpha=0.5)


plt.xlabel('PC1')
plt.ylabel('PC2')
plt.title('PCA of Pitch Profiles Across Measures')
plt.legend(title='Year Bin', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

## UMAP

```python
reducer = umap.UMAP(n_neighbors=10, min_dist=0.01, random_state=42)
X_umap = reducer.fit_transform(X_scaled)

plt.figure(figsize=(8,6))

colors = plt.cm.get_cmap('tab10', len(bins))  

for i, year in enumerate(bin_labels):
    subset = X_umap[notes_no_dups['year_bin'] == year]

    n_sample = int(len(subset) * 0.3)

    subset = subset[np.random.choice(len(subset), n_sample, replace=False)]

    plt.scatter(subset[:,0], subset[:,1], color=colors(i), label=str(year), alpha=0.5)

plt.xlabel('UMAP1')
plt.ylabel('UMAP2')
plt.title('UMAP of Pitch Profiles Across Measures')
plt.legend(title='Year Bin', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

## PCA + UMAP + HDBSCAN

```python
pca = PCA(n_components=22)
X_pca = pca.fit_transform(X_scaled)

print("Explained variance ratio:", pca.explained_variance_ratio_)
print("Total variance explained by 22 components:", sum(pca.explained_variance_ratio_))
```

```python
# 1. Cluster on PCA data
clusterer = hdbscan.HDBSCAN(min_cluster_size=20, min_samples=10)
labels = clusterer.fit_predict(X_pca)  # cluster on X_scaled

# 2. Attach cluster labels to df
notes_no_dups['cluster'] = labels

# 3. Compute UMAP for visualization
reducer = umap.UMAP(n_neighbors=10, min_dist=0.01, random_state=42)
X_umap = reducer.fit_transform(X_pca)
```

```python
cluster_sizes = np.bincount(labels[labels >= 0])
big_clusters = np.where(cluster_sizes > 50)[0]

# Colour noisy points grey 
colors = np.tile(np.array([0.83, 0.83, 0.83, 1.0]), (len(labels), 1))  # light gray RGBA

# Colour for clusters
palette = plt.cm.tab20(np.linspace(0, 1, len(big_clusters)))

# Assign colours to clusters
for i, cl in enumerate(big_clusters):
    colors[labels == cl] = palette[i]

# Plot
fig, ax = plt.subplots(figsize=(8, 6))
scatter = ax.scatter(X_umap[:, 0], X_umap[:, 1], c=colors, s=5, alpha=0.6)

if len(big_clusters) > 0:
    cmap = ListedColormap([palette[i] for i in range(len(big_clusters))])
    norm = BoundaryNorm(np.arange(len(big_clusters)+1)-0.5, len(big_clusters))
    sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
    sm.set_array([])
    cbar = fig.colorbar(sm, ax=ax, ticks=np.arange(len(big_clusters)))
    cbar.ax.set_yticklabels(big_clusters)
    cbar.set_label("Cluster ID")

ax.set_title("UMAP visualization colored by HDBSCAN clusters")
ax.set_xlabel("UMAP1")
ax.set_ylabel("UMAP2")
plt.show()
```

```python
x = np.arange(-16, 20)

for big_cluster in big_clusters:

    subset = notes_no_dups[notes_no_dups['cluster'] == big_cluster]
    subset = subset.drop(columns=['cluster', 'mc', 'recording_year', 'artist', 'year_bin'])
    
    plt.figure(figsize=(10,3))
    
    for _, row in subset.iterrows():
        plt.plot(x, row.values, alpha=0.3)
    
    avg_vector = subset.mean(axis=0)
    plt.plot(x, avg_vector.values, color="black", linewidth=2, label="Average")

    xtick_labels = [pa.get_note(val) for val in x]
    
    plt.xticks(ticks=x, labels=xtick_labels, rotation=45)
    plt.xlabel("TPC")
    plt.ylabel("Frequency per measure")
    plt.title(f"Cluster {big_cluster}")
    plt.legend()
    plt.show()
```
