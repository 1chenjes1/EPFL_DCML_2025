```python
from music21 import harmony

def label_to_pcs(label):
    cs = harmony.ChordSymbol(label)
    pcs = sorted([p.pitchClass for p in cs.pitches])
    return pcs
```

```python
def generate_pcs(df):

    pitchclass = df.groupby(by=['time', 'fnames']).agg({
        'mc': 'first',
        'mn': 'first',
        'mc_onset': 'first',
        'duration': 'first',
        'note': lambda x: list(set(x))  # collect unique notes as list
    }).reset_index()

    return pitchclass
```
