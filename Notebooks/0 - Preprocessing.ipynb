```python
def groupby_standard(df, length=3):
    # Step 1: Get unique (fnames, workTitle) pairs
    ref = df[['fnames', 'workTitle']].drop_duplicates()

    # Step 2: Group by workTitle → count how many files (fnames) per title
    ref_count = ref.groupby('workTitle').size().reset_index(name='count')

    # Step 3: Keep only titles with ≥ length files
    ref_filtered = ref_count[ref_count['count'] >= length]

    # Step 4: Filter big df to only those titles
    title_grouped = df[df['workTitle'].isin(ref_filtered['workTitle'])]

    ref_filtered = ref_filtered.sort_values(by='count', ascending=False)

    return title_grouped, ref_filtered

title_grouped, ref = groupby_standard(df, length=3)
ref
```

```python
def groupby_artist(df, length=3):
    # Step 1: Get unique (fnames, artist) pairs
    ref = df[['fnames', 'artist']].drop_duplicates()

    # Step 2: Group by artist → count how many files (fnames) per title
    ref_count = ref.groupby('artist').size().reset_index(name='count')

    # Step 3: Keep only titles with ≥ length files
    ref_filtered = ref_count[ref_count['count'] >= length]

    # Step 4: Filter big df to only those titles
    artist_grouped = df[df['artist'].isin(ref_filtered['artist'])]

    ref_filtered = ref_filtered.sort_values(by='count', ascending=False)

    return artist_grouped, ref_filtered

artist_grouped, ref = groupby_artist(df, length=3)
ref
```

```python
def groupby_year(df, length=3):
    # Step 1: Get unique (fnames, artist) pairs
    ref = df[['fnames', 'recording_year']].drop_duplicates()

    # Step 2: Group by artist → count how many files (fnames) per title
    ref_count = ref.groupby('recording_year').size().reset_index(name='count')

    # Step 3: Keep only titles with ≥ length files
    ref_filtered = ref_count[ref_count['count'] >= length]

    # Step 4: Filter big df to only those titles
    year_grouped = df[df['recording_year'].isin(ref_filtered['recording_year'])]

    ref_filtered = ref_filtered.sort_values(by='count', ascending=False)

    return year_grouped, ref_filtered

year_grouped, ref = groupby_year(df, length=3)
ref
```

```python
# Group by artist, aggregate min and max
df['recording_year'] = pd.to_numeric(df['recording_year'], errors='coerce')

agg_df = (
    df.groupby('artist').agg(
        min_year=('recording_year', 'min'),
        max_year=('recording_year', 'max'),
        count=('workTitle', 'nunique')
    ).reset_index()
)
agg_df['year_range'] = agg_df['min_year'].astype(str) + "–" + agg_df['max_year'].astype(str)

agg_df[['artist', 'year_range', 'count']].sort_values(by='count', ascending=False).head(10)
```
